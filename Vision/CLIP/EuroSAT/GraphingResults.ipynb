```python
import pandas as pd
import matplotlib.pyplot as plt 
import os
import numpy as np
```

```python
results = [] # File Loading

path = '/Users/darrinobrien/Desktop/School/Summer_Programs/Ongoing/Algoverse/Vision/CLIP/EuroSAT/Results/Entire_Transformation_Matrix_W'

try:
    for filename in os.listdir(path):
        if filename == ".DS_Store":
            continue
        file_path = os.path.join(path, filename)
        if os.path.isfile(file_path):
            results.append(file_path)
except FileNotFoundError:
    print(f"Error: The Folder '{path}' was not found.")
except Exception as e:
    print(f"An error occured: {e}")
```

```python
# All Accuracy Together
train_size = []
for i in range(38):
    train_size.append(data[i].loc[0, "Train_Data_Size"])
acc_vals = {}
for i in range(12):
    acc_vals[i] = []
for i in range(12):
    for j in range(38):
        acc_vals[i].append(data[j].loc[i, "Accuracy"])

colors = plt.cm.tab20(np.linspace(0,1,12))
plt.figure(figsize=(20,12))
for i in range(12):
    plt.plot(train_size, acc_vals[i], marker="o", color=colors[i], label=f"{i} Layer Accuracy")
plt.plot([0, 1400], [0.9819, 0.9819], label="Fine-Tuned Accuracy")
plt.plot([0, 1400], [0.1148, 0.1148], label="Base Accuracy")

plt.legend(
    loc="center left",
    bbox_to_anchor=(1.0, 0.5)
)
plt.title("Augmented CLIP-EuroSAT with W at All Layers - Classification Accuracy")
plt.xlabel("Size of training Data", labelpad=6)
plt.tight_layout()
plt.savefig(f'/Users/darrinobrien/Desktop/School/Summer_Programs/Ongoing/Algoverse/Vision/CLIP/EuroSAT/Graphed_Results/All_Augmented_Layers_Classifiction_Accuracy', dpi=600)
plt.show()
```
