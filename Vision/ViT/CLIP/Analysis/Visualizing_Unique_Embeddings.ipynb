```python
from datasets import load_dataset
from collections import defaultdict
import matplotlib.pyplot as plt
```

```python
dataset = load_dataset('ylecun/mnist') # https://huggingface.co/datasets/ylecun/mnist
num_classes = 10
domain = "geometric_abberations"
dataset_name = "MNIST"
size_nums = [i for i in range(1, 100, 5)]
size_nums.insert(0, float('inf'))
```

```python
split = dataset["train"].train_test_split(test_size=0.2, seed=66)

train = split["train"]
val = split["test"]
test = dataset["test"]
```

```python
train_size = 0

labels = train["label"]

label_to_indices = defaultdict(list)

for idx, label in enumerate(labels):
    label = int(label)
    label_to_indices[label].append(idx)

filtered_train = {
    label: train.select(indices) for label, indices in label_to_indices.items()
}

for label, ds in filtered_train.items():
    print(f"Label {label}: {len(ds)} examples")
```

```python
special_split = {}
for i in range(10):
    special_split[i] = []
    for j in range(100):
        special_split[i].append(label_to_indices[i][j])
```

```python
weird_img = {}
weird_img[0] = [2,85,45,55,19,26,1,52,84]
weird_img[9] = [66,83,35,60,98,56,89,66,70,45,62,79]
```

```python
for i in weird_img[9]:
    indice = special_split[9][i]
    plt.imshow(train[indice]["image"], cmap="gray")
    plt.title(f"Digit 9 Train Indice {indice} | First 100 Train Indice {i}")
    plt.savefig(f"../Embedding_Graphs/{dataset_name}/Base_Fine_Tuned/{domain}/Digit_9/Img_Num_{i}")
    plt.show()
```
